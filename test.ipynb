```python
import pandas as pd
from rouge_score import rouge_scorer
from tqdm import tqdm
import os
import sys

# Add parent dir to path to import rag_pipeline
os.path.abspath(os.path.join(os.getcwd(), "data"))

from app.rag_pipeline import build_qa_chain

data = pd.read_csv("doc.csv")
scorer = rouge_scorer.RougeScorer(['rouge1', 'rougeL'], use_stemmer=True)
qa_chain = build_qa_chain()
results = []

for idx, row in tqdm(data.iterrows(), total=len(data)):
    question = row["question"]
    reference = row["reference"]

    try:
        generated = qa_chain({"query": question})['result']
    except Exception as e:
        generated = "ERROR"
        print(f"Error on Q{idx+1}: {e}")

    scores = scorer.score(reference, generated)
print(type(scores.reference))
```
